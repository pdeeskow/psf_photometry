# Scatterplot: v_mag_cal vs jd
Dieses Notebook lädt `results_targets.csv` und erstellt einen Scatterplot der Spalte `v_mag_cal` gegen `jd`.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('results_targets.csv')

x = df['jd']
y = df['v_mag_cal']

# Robust plot range: ignore extreme outliers via quantiles
x_min, x_max = x.quantile([0.03, 0.97])
y_min, y_max = y.quantile([0.03, 0.97])

plt.figure(figsize=(8, 5))
plt.scatter(x, y, s=20, alpha=0.8)
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.xlabel('jd')
plt.ylabel('v_mag_cal')
plt.title('v_mag_cal vs jd (outliers ignored in axis range)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
